```python
import pandas as pd
```

```python
#for joinging handpicked images from two different dataset
#read them
df_HS_HP = pd.read_csv("handpicked_HS_Img.csv")
df_NHS_HP = pd.read_csv("handpicked_NHS_Img.csv")

#concatenate them
#and shuffle the rows
df_HP_C = pd.concat([df_HS_HP, df_NHS_HP], ignore_index=True)
df_HP_C = df_HP_C.sample(frac=1).reset_index(drop=True)

#save the concatenated dataframe to a new CSV file
df_HP_C.to_csv("HP_Img.csv", index=False)
df_HP_C.head()
```

```python
#for joinging Dalle images from two different dataset
#read them
df_DE = pd.read_csv("DE_Img.csv")
df_SD = pd.read_csv("SD_Img.csv")
df_HP = pd.read_csv("HP_Img.csv")

#concatenate them
#and shuffle the rows
df_HP_C = pd.concat([df_DE, df_SD, df_HP], ignore_index=True)
df_HP_C = df_HP_C.sample(frac=1).reset_index(drop=True)

#save the concatenated dataframe to a new CSV file
df_HP_C.to_csv("Complete_Img.csv", index=False)
df_HP_C.head()
```
