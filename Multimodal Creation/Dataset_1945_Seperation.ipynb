```python
import pandas as pd
```

```python
#load the datset
df = pd.read_csv("hope_speech.csv",sep=',', names=['Text', 'Class'])

#this will split the dataset into handpicked and generated grabbing first 648 into handpicked and rest into generated
handpicked_HS = df.iloc[:648].copy()
generate_HS = df.iloc[648:].copy()

#save the datasets
handpicked_HS.to_csv("handpicked_HS.csv", index=False)
generate_HS.to_csv("generated_HS.csv", index=False)

#display to show the seperation
print(df.head(650))
print(handpicked_HS.head())
print(generate_HS.head())
```

```python
#load the datset
df = pd.read_csv("generated_HS.csv",sep=',', names=['Text', 'Class'])

#this will split the dataset into handpicked and generated grabbing first 648 into handpicked and rest into generated
SD_Gen_HS = df.iloc[:649].copy()
DE_Gen_HS = df.iloc[649:].copy()

#save the datasets
SD_Gen_HS.to_csv("SD_Gen_HS.csv", index=False)
DE_Gen_HS.to_csv("DE_Gen_HS.csv", index=False)

#display to show the seperation
print(df.head(650))
print(SD_Gen_HS.head())
print(DE_Gen_HS.head())
```
