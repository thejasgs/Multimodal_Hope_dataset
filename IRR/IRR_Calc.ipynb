```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score

an1 = pd.read_csv('Combined_Images_IRR_A1.csv')
an2 = pd.read_csv('Combined_Images_IRR_A2.csv')

combined = pd.merge(an1[['image', 'An1']], 
                  an2[['image', 'An2']], 
                  on='image', 
                  how='inner')
combined['An1'].astype(int)
combined['An2'].astype(int)
combined.to_csv("combined_IRR_AN1_AN2.csv", index=False)

```

```python
df = pd.read_csv("combined_IRR_AN1_AN2.csv")

anno1 = df["An1"]
anno2 = df["An2"]

kappa = cohen_kappa_score(anno1, anno2)

print("Cohen Kappa:", kappa)
```
